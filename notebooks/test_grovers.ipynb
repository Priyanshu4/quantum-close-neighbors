```python
from quantum_close_neighbors.grovers import AbstractGroverResult
from quantum_close_neighbors.grovers.probabilistic_grovers import ProbabilisticGrovers
from quantum_close_neighbors.grovers.qiskit_grovers import QiskitGroversPhaseOracle
from tqdm import tqdm
```

```python
# Note that n_elements should be a power of 2 for Qiskit grovers and probabilistic grovers to produce the same results
# This is because the real number of elements in Qiskit simulation is always 2^q, where q is the number of qubits

n_elements = 16
n_marked_elements = 10
num_iterations = 0
samples = 10
trials = 100
p_estimate_shots = 100
```

```python
prob_grovers_marked_seen = []
prob_grovers_marked_unseen = []
prob_grovers_unmarked = []

for i in tqdm(range(trials)):
    marked_seen = 0
    marked_unseen = 0
    unmarked = 0

    grovers = ProbabilisticGrovers(n_elements, n_marked_elements)
    for _ in range(samples):
        result = grovers.run(num_iterations)

        if result == AbstractGroverResult.MARKED_SEEN:
            marked_seen += 1
        elif result == AbstractGroverResult.MARKED_UNSEEN:
            marked_unseen += 1
        else:
            unmarked += 1

    prob_grovers_marked_seen.append(marked_seen)
    prob_grovers_marked_unseen.append(marked_unseen)
    prob_grovers_unmarked.append(unmarked)
```

```python
qiskit_grovers_marked_seen = []
qiskit_grovers_marked_unseen = []
qiskit_grovers_unmarked = []  

for i in tqdm(range(trials)):
    marked_seen = 0
    marked_unseen = 0
    unmarked = 0

    grovers = QiskitGroversPhaseOracle.init_with_random_marked_elements(n_elements, n_marked_elements)
    for _ in range(samples):
        result = grovers.run(num_iterations)

        if result == AbstractGroverResult.MARKED_SEEN:
            marked_seen += 1
        elif result == AbstractGroverResult.MARKED_UNSEEN:
            marked_unseen += 1
        else:
            unmarked += 1

    qiskit_grovers_marked_seen.append(marked_seen)
    qiskit_grovers_marked_unseen.append(marked_unseen)
    qiskit_grovers_unmarked.append(unmarked)
```

```python
print("Probabilistic Grover's Algorithm")
if trials > 0:
    print("Average marked seen: {}".format(sum(prob_grovers_marked_seen) / trials))
    print("Average marked unseen: {}".format(sum(prob_grovers_marked_unseen) / trials))
    print("Average unmarked: {}".format(sum(prob_grovers_unmarked) / trials))
print(f"Expected p_success: {ProbabilisticGrovers(n_elements, n_marked_elements).p_success(num_iterations)}")


print("Qiskit Grover's Algorithm")
if trials > 0:
    print("Average marked seen: {}".format(sum(qiskit_grovers_marked_seen) / trials))
    print("Average marked unseen: {}".format(sum(qiskit_grovers_marked_unseen) / trials))
    print("Average unmarked: {}".format(sum(qiskit_grovers_unmarked) / trials))

print("Estimated p_success: {}".format(QiskitGroversPhaseOracle.init_with_random_marked_elements(n_elements, n_marked_elements).estimate_p_success(num_iterations, p_estimate_shots, with_tqdm=True)))
```
